```python
# Import Libraries
import pandas as pd
import numpy as np

# Display all columns
pd.set_option("display.max_columns", None)

# Display wider tables
pd.set_option("display.width", 200)
```

```python
# Load the Excel Workbook
FILE_PATH = "../data/raw/Basic Course Score.xlsx"

# Read workbook
excel_file = pd.ExcelFile(FILE_PATH)

# Display sheet names
print(excel_file.sheet_names)
```

```python
# Read Every Sheet
march = pd.read_excel(FILE_PATH, sheet_name="March")
april = pd.read_excel(FILE_PATH, sheet_name="April")
may = pd.read_excel(FILE_PATH, sheet_name="May")
june = pd.read_excel(FILE_PATH, sheet_name="June")
```

```python
# Check Dataset Size
datasets = {
    "March" : march,
    "April" : april,
    "May" : may,
    "Jun" : june,
}

for month, df in datasets.items():
    print(f"{month}")
    print(f"Rows    : {df.shape[0]}")
    print(f"Columns : {df.shape[1]}")
    print("-" * 40)
```

```python
# View First Few Rows
# march.head()
# april.head()
# may.head()
june.head()
```

```python
# View Column Names
for month, df in datasets.items():
    print(f"\n{month}")
    print(df.columns.tolist())
```

```python
# Data Types
for month, df in datasets.items():
    print(f"\n{month}")
    print(df.dtypes)
```

```python
# Missing Values
for month, df in datasets.items():
    print(f"\n======= {month} =======")
    print(df.isnull().sum())
```

```python
# Summary Statistics
for month, df in datasets.items():
    print(f"\n===== {month} =====")
    print(df.describe(include="all"))
```
