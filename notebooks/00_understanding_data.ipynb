## Dataset Overview

The dataset consists of four monthly student reports.

- March 2026
- April 2026
- May 2026
- June 2026

Each worksheet contains one month's student performance.

```python
# Import Libraries
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 200)
```

```python
# Load Dataset
FILE_PATH = "../data/raw/Basic Course Score.xlsx"

excel = pd.ExcelFile(FILE_PATH)

excel.sheet_names
```

```python
# Inspect Worksheets
datasets = {}

for sheet in excel.sheet_names:
    datasets[sheet] = pd.read_excel(FILE_PATH, sheet_name=sheet)

for month, df in datasets.items():
    print("=" * 100)
    print(month)
    print("=" * 100)

    display(df.head())
```

```python
# Explore Dataset Structure
for month, df in datasets.items():
    print("=" * 50)
    print(month)
    print("=" * 50)

    print(f"Shape: {df.shape}")

    print()

    df.info()
```

```python
# Missing Values
for month, df in datasets.items():
    print("=" * 25)
    print(month)
    print("=" * 25)

    display(df.isnull().sum())
```

```python
# Duplicate Records
for month, df in datasets.items():
    print("=" * 25)
    print(month)
    print("=" * 25)

    print(df.duplicated().sum())
```

## Initial Findings

- The workbook contains four monthly reports.
- Each worksheet represents one month.
- The number of students decreases over time.
- The assessment structure changes between months.
- Most students are male.
- The Comment column contains mostly blank values.
- No duplicate records were detected.
