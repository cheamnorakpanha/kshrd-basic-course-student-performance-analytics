#### Import Libraries

The following libraries are used throughout this notebook.
- pandas
- numpy

```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 200)
```

#### Load Dataset

- The monthly student performance reports are stored in a single Excel workbook.
- This section loads the workbook and identities the available worksheets.
- Each worksheet represents one month of student performance data.

```python
FILE_PATH = "../data/raw/Basic Course Score.xlsx"

excel = pd.ExcelFile(FILE_PATH)

excel.sheet_names
```

#### Inspect Worksheets

- Before performing any analysis, it is important to inspect the structure of each worksheet.
- Displaying the first few records allows us to verify that the data has been loaded correctly and provides an initial understanding of the available information.

```python
datasets = {}

for sheet in excel.sheet_names:
    datasets[sheet] = pd.read_excel(FILE_PATH, sheet_name=sheet)

for month, df in datasets.items():
    print("=" * 80)
    print(month)
    print("=" * 80)

    display(df.head(3))
```

#### Dataset Structure

- Understanding the structure of the dataset helps identify the number of observations, available attributes, and potential differences between monthly reports.

- This section summarizes the number of rows, columns, missing values, and duplicate records for each worksheet.

```python
summary = []

for month, df in datasets.items():
    summary.append({
        "Month": month,
        "Rows": df.shape[0],
        "Columns": df.shape[1],
        "Missing Values": df.isna().sum().sum(),
        "Duplicate Rows": df.duplicated().sum()
    })

summary_df = pd.DataFrame(summary)

summary_df
```

#### Data Types

- Each column has an associated data type that determines how it should be processed.
- Verifying data types at an early stage helps identify potential issues such as numeric values stored as text or incorrectly formatted columns.- 

```python
for month, df in datasets.items():
    print(f"\n===== {month} =====")

    df.info()
```

#### Missing Values

- Missing values are common in real-world datasets.
- This section identifies columns containing missing values so that appropriate cleaning strategies can be determined during the data preparation stage.

```python
for month, df in datasets.items():
    print("=" * 25)
    print(month)
    print("=" * 25)

    display(df.isnull().sum())
```

#### Duplicate Records

- Duplicate records may introduce bias into the analysis.
- This section checks whether any duplicate rows exist within each monthly report.

```python
for month, df in datasets.items():
    print("=" * 25)
    print(month)
    print("=" * 25)

    print(df.duplicated().sum())
```

#### Summary Statistics

- Descriptive statistics provide a quick overview of the numerical variables in the dataset.
- Key measures such as the mean, standard deviation, minimum, and maximum values help us understand the distribution of student scores.

```python
for month, df in datasets.items():
    display(df.describe())
```

```python
categorical_columns = [
    "Gender",
    "Class",
    "Comment"
]

for month, df in datasets.items():
    print("=" * 25)
    print(month)
    print("=" * 25)

    for col in categorical_columns:

        print(f"\n{col}")

        print(df[col].value_counts(dropna=False))
```

## Key Findings

From the initial inspection, the following observations were made:

- The workbook contains four monthly worksheets.
- Each worksheet represents one month of student performance.
- The number of students changes slightly across months.
- The assessment structure is not identical across all months.
- Most missing values occur in the Comment column.
- No duplicate records were identified.
