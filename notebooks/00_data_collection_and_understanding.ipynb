# KSHRD Basic Course Student Performance Analytics

## 00. Data Collection & Understanding

This notebook documents the collection and initial understanding of the
KSHRD Basic Course student performance dataset.

The dataset contains monthly student results from March to June 2026.
The purpose of this project is to apply descriptive analytics to identify
student performance patterns, monthly trends, class performance, subject
performance, attendance patterns, and student status.

### Import Libraries

```python
import pandas as pd
import numpy as np
from pathlib import Path
```

### Define the Raw Data Path

```python
RAW_DATA_PATH = Path("../data/raw/Basic Course Score.xlsx")

RAW_DATA_PATH.exists()
```

### Load the Workbook

```python
excel_file = pd.ExcelFile(RAW_DATA_PATH)

excel_file.sheet_names
```

### Load Each Sheet

```python
sheets = {}

for sheet in excel_file.sheet_names:
    sheets[sheet] = pd.read_excel(RAW_DATA_PATH, sheet_name=sheet)

sheets.keys()
```

### Initial Dataset Overview

```python
for month, df in sheets.items():
    print(f"{month}: {df.shape}")
```

### Inspect Each Sheet

```python
for month, df in sheets.items():
    print(f"\n{'='*50}")
    print(f"{month}")
    print(f"{'='*50}")
    display(df.head())
```

```python
for month, df in sheets.items():
    print(f"\n{month}")
    print(df.columns.tolist())
```

## Data Source

The dataset was collected from the monthly student result reports of the
KSHRD Basic Course for March, April, May, and June 2026.

The original Excel workbook contains one worksheet for each reporting month.
The raw workbook is preserved in its original form under `data/raw/` and is
used as the source for subsequent data cleaning and preparation.

### Initial Dataset Summary

```python
summary = []

for month, df in sheets.items():
    summary.append(
        {
            "Month": month,
            "Rows": df.shape[0],
            "Columns": df.shape[1],
            "Missing Values": df.isna().sum().sum(),
            "Duplicate Rows": df.duplicated().sum(),
        }
    )

collection_summary = pd.DataFrame(summary)

collection_summary
```

### Missing Values by Column

This section identifies missing values in each monthly dataset. The purpose is to understand where missing data occurs before making any cleaning or imputation decisions.

```python
for month, df in sheets.items():
    print(f"\n{'=' * 60}")
    print(month)
    print('=' * 60)

    missing = df.isna().sum()
    missing = missing[missing > 0].sort_values(ascending=False)

    display(missing.to_frame("Missing Values"))
```

### Duplicate Student Names

This section checks whether multiple records within the same month have the same student name. This is important because student names may later be used to track performance across different months.

```python
for month, df in sheets.items():
    duplicates = df[df["Name"].duplicated(keep=False)].sort_values("Name")

    print(f"\n{month}: {len(duplicates)} rows with duplicated names")

    if not duplicates.empty:
        display(
            duplicates[["Rank", "Name", "Gender", "Class"]]
        )
```

### Student Status Distribution

The `Comment` column contains student status information:

- `CL` = Class Leader
- `W` = Warning
- `2W` = Two Warnings
- Blank = No recorded status

```python
for month, df in sheets.items():
    print(f"\n{'=' * 60}")
    print(month)
    print('=' * 60)

    display(
        df["Comment"]
        .fillna("No Status")
        .value_counts()
        .to_frame("Count")
    )
```

### Data Types

This section examines the data types of the variables in each monthly dataset. This helps identify columns that may require conversion during the data cleaning and preparation stage.

```python
for month, df in sheets.items():
    print(f"\n{'=' * 60}")
    print(month)
    print('=' * 60)

    display(df.dtypes.to_frame("Data Type"))
```
